### Transforming data (e.g., creating new columns, applying functions to columns)

Create a new DataFrame with transformed data

```python
# OPTION 1:
df_orders_new = df_orders.assign(
    is_weekend = df_orders["order_dow"].isin([0, 6]),
    is_morning = df_orders["order_hour_of_day"] < 12
)
```

```python
#OPTION 2:
df_orders["is_weekend"] = df_orders["order_dow"].isin([0, 6])
df_orders["is_morning"] = df_orders["order_hour_of_day"] < 12

df_orders.head()
```

Creating an order frequency category based on order_number

```python
df_orders["order_frequency_category"] = df_orders["order_number"].apply(
    lambda x: "New" if x == 1 
    else "Low" if x <= 5 
    else "High"
)

df_orders.head()
```

Handling missing values in days_since_prior_order

```python
df_orders["days_since_prior_order"] = df_orders["days_since_prior_order"].fillna(0)

df_orders.head()
```

Creating time-of-day buckets

```python
df_orders["time_bucket"] = df_orders["order_hour_of_day"].apply(
    lambda x: "Morning" if 6 <= x < 12
    else "Afternoon" if 12 <= x < 18
    else "Evening"
)

df_orders.head()
```

### Transponsing a DateFrame

```python
df_departments = pd.read_csv("../data/raw/departments.csv")
df_departments.head()
```

```python
df_departments.T
```

```python
# The Flip
# This flips the rows and columns.If your departments were columns,they are now rows.

df_dep_t = df_departments.T
```

```python
# Bringing the Labels into the Table
# By default, the old column names are stuck in the "Index" (the far left margin). 
# This command pushes them into the actual data area so we can work with them.
df_dep_t.reset_index(inplace=True)
```

```python
# Grabbing the New Headers
# We take the very first row of the data (iloc[0]) and save it into a variable. 
# We intend for this row to become the names of our columns.
new_header = df_dep_t.iloc[0]
```

```python
# Trimming the Data
# Since the first row is now saved as our "header," we don't want it duplicated inside the table. 
# This line creates a new DataFrame that starts from the second row (1:) onwards.
df_dep_t_new = df_dep_t[1:]
```

```python
# Assigning the Names
# Finally, we tell Pandas: "Use the values we saved in new_header as the official titles for the columns."
df_dep_t_new.columns = new_header   

df_dep_t_new.head()
```

Why go through all this trouble?

In datasets like the one you're using, "Departments" might be stored as a single horizontal row. To join this data with your orders DataFrame, you need "Department" to be a vertical column so Pandas can match the IDs.

# Understanding Relative Paths in Python

The most practical "life skills" in programming: 

File Paths. Understanding this is what ensures your code works on your computer, your friend's computer, and the cloud without breaking.

A file path is just a set of directions. Here is the breakdown of the two types mentioned:

1. Absolute Path (The "Full Address")
This is the equivalent of a full mailing address, including the city, state, and zip code. It starts from the very "root" of your hard drive.

Example: C:/Users/YourName/Project/data/file.xlsx

The Problem: If you send this code to a teammate, it will crash. Why? Because their computer doesn't have a user named "YourName." Absolute paths are rigid and "fragile."

2. Relative Path (The "Directions from Here")
This is like telling someone, "Go down the hall and take the first door on the right." It only works if you both start in the same place (the Current Working Directory).

Example: data/file.xlsx

The Benefit: This is portable. As long as your teammate has the same project folder structure as you, the code will work perfectly for them. This is the "industry standard" for collaborative projects.

3. How to "Navigate" with Relative Paths
You’ll often see special symbols in relative paths that help you move through folders:

./ : Means "Look in the current folder."

../ : Means "Go up one folder" (the parent directory).

Real Example:

In previous step, the used path for exporting data:
"../data/processed/orders_cleaned.csv"

This tells Python:

../ : Go out of the current folder where my Jupyter Notebook is.

data/ : Look for a folder called "data."

processed/ : Look for a folder inside that called "processed."

orders_cleaned.csv : Save the file here.

## Option A

### The os.path

Windows uses backslashes (\) for paths, while Mac and Linux use forward slashes (/). 

If you hardcode your paths with /, your code might crash on a teammate's Windows machine.

os.path.join() solves this. It detects which computer you are using and automatically puts the correct slashes in between your folders.

```python
import os

# Get current working directory
print("Current working directory:", os.getcwd())

# Join a relative path
file_path = os.path.join("data", "file.xlsx")
print("Relative path:", file_path)
```

Finding Your "Home Base"

os.getcwd() stands for Get Current Working Directory.
When you run this in a Jupyter Notebook, it tells you exactly where the computer thinks it is sitting. 
This is vital for debugging! If your file won't load, run this command to see if you are actually in the folder you think you're in.

If you are working inside main.ipynb (which is in the scripts or notebooks folder) and you want to open sample.xlsx (which is in the data folder), you have to tell Python to:

Leave the current room: .. (This takes you up to the main project/ folder).

Go into a new room: data/

Find the file: sample.xlsx

Final Path: ../data/sample.xlsx

## Option B

### pathlib is the "Modern Pro" Choice

```python
import pandas as pd
from pathlib import Path

# Define relative path to the data folder from scripts/
file_path = Path("../data") / "sample.csv"

# Read the CSV file
df = pd.read_csv(file_path)

# Display the first few rows
print(df.head())
```

The Magic Forward Slash (/):

Notice this line: file_path = Path("../data") / "sample.csv"

In standard Python, you can't divide two pieces of text.

But with pathlib, the / acts as a "Joiner." 

It automatically handles the folder slashes for you, making the code very readable and cross-platform (Windows, Mac, and Linux).
