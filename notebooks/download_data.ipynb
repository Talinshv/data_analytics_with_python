```python
import os
import pandas as pd

URL = "https://raw.githubusercontent.com/hovhannisyan91/tableau-analytics-portfolio/refs/heads/main/session02/data"

tables = [
    "customers.csv",
    "customer_addresses.csv",
    "orders.csv",
    "order_details.csv",
    "products.csv",
    "returns.csv",
    "salespeople.csv"
]

# Path to the tabluae_lab folder (jumping up one level from the notebooks folder)
output_folder = "../data/tabluae_lab"
os.makedirs(output_folder, exist_ok=True)

# Loop and download each file
for table in tables:
    url = f"{URL}/{table}"
    print(f"Downloading {table}...")
    
    # Read the data from GitHub
    df = pd.read_csv(url)
    
    # Save it into the tabluae_lab folder
    df.to_csv(f"{output_folder}/{table}", index=False)

print("✨ Done! Check your data/tabluae_lab folder now.")
```
