# Part 3

Segmentation rules:

- “Premium” if revenue > 250
- “Standard” if 100 < revenue ≤ 250
- “Low” otherwise

```python
# Create a list of segment labels.
segments = ["premium" if r > 250 else "standard" if r > 100 and r <= 250 else "low" for r in revenues]
segments
```

```python
# Count how many customers fall into each segment.
count_customers = [0, 0, 0]
for segment in segments:
    if segment == "premium":
        count_customers[0] = count_customers[0] + 1
    elif segment == "standard":
        count_customers[1] = count_customers[1] + 1
    else:
        count_customers[2] = count_customers[2] + 1
count_customers 
```

Add Markdown explanation: Why does the order of elif statements matter?

Python evaluates conditional statements from top to bottom. Once it finds a condition that is True, it executes that block and skips the rest of the chain.

If we had put the "Standard" check (r > 100) before the "Premium" check (r > 250), a revenue of 310 would trigger the r > 100 condition and be labeled "Standard." Python would never even look at the "Premium" rule. To get accurate results, you must generally order your conditions from the most specific (or highest value) to the most general (or lowest value).

Explain in Markdown why sets are unordered.


Sets are unordered because they are implemented using a data structure called a Hash Table.Unlike a List, which stores items in a specific sequence (like passengers in a row of seats), a Set uses a "hashing function" to assign each element a unique "address" in memory. This allows the computer to find an item almost instantly because it knows the exact address, rather than searching through the whole list. Because the goal is speed and uniqueness rather than sequence, the order in which items appear is not preserved.

Add Markdown explanation: Why are dictionaries useful for structured data?

Dictionaries are essential for structured data because they use key-value pairs, which allow us to label our data. Unlike a list where you have to remember that "index 0 is the name" and "index 1 is the revenue," a dictionary explicitly links a label (the key) to its data (the value).

This makes the data:

Self-describing: It's immediately clear what each piece of information represents.

Efficient: You can look up a specific value instantly using its key without searching through a whole list.

Flexible: You can easily add or remove fields (keys) for one record without breaking the structure of others.

```python
import pandas as pd

data = {
    "revenue": revenues
}

df = pd.DataFrame(data)
df
```

```python
# 1. Add column "revenue_after_tax" (Vectorized math!)
df["revenue_after_tax"] = df["revenue"] * 0.9
```

```python
# 2. Add column "segment" using conditional logic
# This is the Pandas version of your if-else logic
df["segment"] = df["revenue"].apply(lambda x: "High" if x > 200 else "Normal")
```

```python
# 3. Filter rows where revenue > 100
df = df[df["revenue"] > 100]
```

```python
# 4. Remove the "revenue_after_tax" column
df = df.drop(columns=["revenue_after_tax"])
```

```python
# 5. Print the shape of the DataFrame
print(f"DataFrame Shape: {df.shape}")
print(df)
```

Why is this easier than manual loops? (Markdown)

Conciseness: Operations that take 4-5 lines in a manual loop (like multiplying every item by tax) can be done in one line with Pandas using vectorized operations.

Performance: Pandas is built on top of C and NumPy. It performs math on entire blocks of memory at once, making it significantly faster than Python's standard for loops when dealing with large datasets.

Readability: The code describes what you want to do (e.g., df.drop()) rather than how to do it (e.g., "Loop through the list, find the index, remove the item").

Integrated Tools: Features like .shape, .info(), and .describe() provide instant summaries of your data that would require manual code to calculate otherwise.

# mutable and immutable objects

Mutable: Objects that can be changed after they are created (e.g., Lists and Dictionaries). You can add, remove, or swap items in place.

Immutable: Objects that cannot be changed once created (e.g., Tuples, Strings, and Integers). Any "change" actually creates a brand-new object in memory.

# loops VS vectorization

Loops: Python must look at each row one by one, interpret the code, and check types every single time. This is "overhead" that slows things down.

Vectorization: Pandas (via NumPy) performs operations in C, which allows it to process entire blocks of data simultaneously across your computer's memory.

# Boolean logic VS SQL WHERE

In Python, Boolean logic (e.g., value > 120) acts as a filter to decide which items stay and which are dropped.

This is conceptually identical to a SQL WHERE clause, which filters a database table to return only rows that meet specific logical criteria.

# lists VS DataFrames

A DataFrame is essentially a collection of Series (columns) aligned by a common index.

Since a Series is basically a "smarter" version of a List, understanding how to iterate, index, and transform lists is the foundational skill needed to manipulate complex tabular data.
